# Perceptron simple


```python
from neural_network import NeuralNetwork, Layer, TrainingType, ActivationFunction, MetricType
from neural_network.utils import (plot_metric, print_predictions, plot_decision_boundary_history)
import matplotlib.pyplot as plt

# Create a neural network with one layer and one neuron, using the step activation function
and_model = NeuralNetwork()
and_model.add_layer(Layer(n_inputs=2, n_neurons=1, activation=ActivationFunction.STEP))

# Train the network on the AND logic gate dataset
x_train = [[0, 0], [0, 1], [1, 0], [1, 1]]
y_train = [[0], [0], [0], [1]]

# Train the network for 10 epochs with a learning rate of 0.1
history = and_model.train(x_train, y_train, epochs=200, learning_rate=1.0, strategy=TrainingType.STOCHASTIC, verbose=True, metrics=(MetricType.MSE, MetricType.ACCURACY, MetricType.PRECISION))


# Courbes
plot_metric(history, MetricType.MSE, "Perceptron simple AND - Évolution de la MSE")
plot_metric(history, MetricType.ACCURACY, "Perceptron simple AND - Évolution de l'accuracy")
plot_metric(history, MetricType.PRECISION, "Perceptron simple AND - Évolution de la précision")

# Évolution des frontières
plot_decision_boundary_history(history, x_train, y_train, title="Perceptron simple AND - Évolution de la frontière de décision")

# Prédictions
print_predictions(and_model, x_train, y_train, classification=True, threshold=0.5)

```
