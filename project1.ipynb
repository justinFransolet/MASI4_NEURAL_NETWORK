# Perceptron simple


```python
from neural_network import NeuralNetwork, Layer, TrainingType, ActivationFunction, MetricType
import matplotlib.pyplot as plt

# Create a neural network with one layer and one neuron, using the step activation function
and_model = NeuralNetwork()
and_model.add_layer(Layer(n_inputs=2, n_neurons=1, activation=ActivationFunction.STEP))

# Train the network on the AND logic gate dataset
x_train = [[0, 0], [0, 1], [1, 0], [1, 1]]
y_train = [[0], [0], [0], [1]]

# Train the network for 10 epochs with a learning rate of 0.1
history = and_model.train(x_train, y_train, epochs=200, learning_rate=1.0, strategy=TrainingType.STOCHASTIC, verbose=True, metrics=(MetricType.MSE, MetricType.ACCURACY, MetricType.PRECISION))

# Display a graph for mse
mse_values = history.metrics[MetricType.MSE.value]
plt.figure(figsize=(8, 4))
plt.plot(range(1, len(mse_values) + 1), mse_values, label="MSE")
plt.title("Evolution de la MSE")
plt.xlabel("Iteration")
plt.ylabel("MSE")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# Display a graph for accuracy
mse_values = history.metrics[MetricType.ACCURACY.value]
plt.figure(figsize=(8, 4))
plt.plot(range(1, len(mse_values) + 1), mse_values, label="Accuracy")
plt.title("Evolution de accuracy")
plt.xlabel("Iteration")
plt.ylabel("Accuracy")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# Display a graph for precision
mse_values = history.metrics[MetricType.PRECISION.value]
plt.figure(figsize=(8, 4))
plt.plot(range(1, len(mse_values) + 1), mse_values, label="Precision")
plt.title("Evolution de la precision")
plt.xlabel("Iteration")
plt.ylabel("Precision")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# Display a graph for decision boundary evolution
plt.figure(figsize=(8, 6))

points_0_x = []
points_0_y = []
points_1_x = []
points_1_y = []

for x, y in zip(x_train, y_train):
    if y[0] == 0:
        points_0_x.append(x[0])
        points_0_y.append(x[1])
    else:
        points_1_x.append(x[0])
        points_1_y.append(x[1])

plt.scatter(points_0_x, points_0_y, label="Classe 0")
plt.scatter(points_1_x, points_1_y, label="Classe 1")

x1_values = [-0.2, 0.0, 0.5, 1.0, 1.5, 2.0, 2.5, 3.0]

for i, params in enumerate(history._params_history, start=1):
    bias = params["bias"]
    w1 = params["weights"][0]
    w2 = params["weights"][1]

    if abs(w2) > 1e-12:
        x2_values = []
        for x1 in x1_values:
            x2 = -(bias + w1 * x1) / w2
            x2_values.append(x2)

        plt.plot(x1_values, x2_values, label=f"epoch {i}")
    elif abs(w1) > 1e-12:
        x_vertical = -bias / w1
        plt.axvline(x=x_vertical, label=f"epoch {i}")

plt.xlim(-0.2, 3.2)
plt.ylim(-0.8, 1.4)
plt.xlabel("x1")
plt.ylabel("x2")
plt.title("Évolution de la frontière de décision")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# Test the network on the training data
for x in x_train:
    output = and_model.predict(x)
    print(f"Input: {x}, Output: {output}")



```
