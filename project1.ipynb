```python
from neural_network import NeuralNetwork, Layer, TrainingType, ActivationFunction, MetricType
import matplotlib.pyplot as plt

# Create a neural network with one layer and one neuron, using the step activation function
and_model = NeuralNetwork()
and_model.add_layer(Layer(n_inputs=2, n_neurons=1, activation=ActivationFunction.STEP))

# Train the network on the AND logic gate dataset
x_train = [[0, 0], [0, 1], [1, 0], [1, 1]]
y_train = [[0], [0], [0], [1]]

# Train the network for 10 epochs with a learning rate of 0.1
history = and_model.train(x_train, y_train, epochs=6, learning_rate=1.0, strategy=TrainingType.STOCHASTIC, metrics=(MetricType.MSE, MetricType.ACCURACY, MetricType.PRECISION))

# Display a graph for mse
mse_values = history.metrics[MetricType.MSE.value]
plt.figure(figsize=(8, 4))
plt.plot(range(1, len(mse_values) + 1), mse_values, label="MSE")
plt.title("Evolution de la MSE")
plt.xlabel("Iteration")
plt.ylabel("MSE")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# Display a graph for accuracy
mse_values = history.metrics[MetricType.ACCURACY.value]
plt.figure(figsize=(8, 4))
plt.plot(range(1, len(mse_values) + 1), mse_values, label="Accuracy")
plt.title("Evolution de accuracy")
plt.xlabel("Iteration")
plt.ylabel("Accuracy")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# Display a graph for precision
mse_values = history.metrics[MetricType.PRECISION.value]
plt.figure(figsize=(8, 4))
plt.plot(range(1, len(mse_values) + 1), mse_values, label="Precision")
plt.title("Evolution de la precision")
plt.xlabel("Iteration")
plt.ylabel("Precision")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# Test the network on the training data
for x in x_train:
    output = and_model.predict(x)
    print(f"Input: {x}, Output: {output}")



```

```python
from neural_network import PerceptronSimple
import matplotlib.pyplot as plt

# Jeu d'apprentissage : porte logique ET (table 2.1)
x_train = [[0, 0], [0, 1], [1, 0], [1, 1]]
y_train = [0, 0, 0, 1]

# Création du perceptron simple
# 2 entrées, taux d'apprentissage η = 1 comme dans le cours
and_model = PerceptronSimple (n_inputs=2, learning_rate=1.0)

# Entraînement
history = and_model.train(
    x_train,
    y_train,
    epochs=20,
    metrics=(MetricType.MSE, MetricType.ACCURACY, MetricType.PRECISION),
    verbose=True
)

# Paramètres finaux
and_model.print_parameters()

# Test sur les données d'entraînement
print("\nPredictions after training:")
for x, y in zip(x_train, y_train):
    pred = and_model.predict(x)
    print(f"Input: {x}, Output: {pred}, Expected: {y}")

# Courbe MSE
mse_values = history.metrics.get(MetricType.MSE.value, [])
if mse_values:
    plt.figure(figsize=(8, 4))
    plt.plot(range(1, len(mse_values) + 1), mse_values, label="MSE")
    plt.title("Évolution de la MSE")
    plt.xlabel("Époque")
    plt.ylabel("MSE")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()

# Courbe Accuracy
accuracy_values = history.metrics.get(MetricType.ACCURACY.value, [])
if accuracy_values:
    plt.figure(figsize=(8, 4))
    plt.plot(range(1, len(accuracy_values) + 1), accuracy_values, label="Accuracy")
    plt.title("Évolution de l'accuracy")
    plt.xlabel("Époque")
    plt.ylabel("Accuracy")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()

# Courbe Precision
precision_values = history.metrics.get(MetricType.PRECISION.value, [])
if precision_values:
    plt.figure(figsize=(8, 4))
    plt.plot(range(1, len(precision_values) + 1), precision_values, label="Precision")
    plt.title("Évolution de la precision")
    plt.xlabel("Époque")
    plt.ylabel("Precision")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()
```
