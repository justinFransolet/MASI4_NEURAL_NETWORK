# Part 1 : AND (descente du gradient / ADALINE)

```python
from neural_network import NeuralNetwork, Layer, TrainingType, ActivationFunction, MetricType
from neural_network.utils import (plot_metric, plot_decision_boundary, print_predictions)

x_train = [[0, 0], [0, 1], [1, 0], [1, 1]]
y_train = [[0], [0], [0], [1]]

and_model = NeuralNetwork()
and_model.add_layer(Layer(n_inputs=2, n_neurons=1, activation=ActivationFunction.IDENTITY))

history = and_model.train(
    x_train=x_train,
    y_train=y_train,
    epochs=50,
    learning_rate=0.1,
    strategy=TrainingType.STOCHASTIC,
    verbose=True,
    metrics=(
        MetricType.MSE,
        MetricType.ACCURACY,
        MetricType.PRECISION,
        MetricType.RECALL,
        MetricType.F1_SCORE
    ),
    update_on_error_only=False,
    mse_threshold=0.001,
    classification_threshold=0.5
)

plot_metric(history, MetricType.MSE, "AND - Évolution de la MSE")
plot_metric(history, MetricType.ACCURACY, "AND - Évolution de l'accuracy")
plot_decision_boundary(and_model, x_train, y_train, "AND - Frontière de décision finale", threshold=0.5)
print_predictions(and_model, x_train, y_train, classification=True, threshold=0.5)
```

# Part 2 : Classification linéairement séparable (table 2.9)

```python
from neural_network import NeuralNetwork, Layer, TrainingType, ActivationFunction, MetricType
from neural_network.utils import (load_csv_dataset, plot_metric, plot_decision_boundary, print_predictions)

x_lin, y_lin = load_csv_dataset("data/table_2_9.csv", target_columns=1, has_header=False)
y_lin = [[0.0] if y[0] == -1.0 else [1.0] for y in y_lin]

lin_model = NeuralNetwork()
lin_model.add_layer(Layer(n_inputs=len(x_lin[0]), n_neurons=1, activation=ActivationFunction.IDENTITY))

history_lin = lin_model.train(
    x_train=x_lin,
    y_train=y_lin,
    epochs=500,
    learning_rate=0.001,
    strategy=TrainingType.STOCHASTIC,
    verbose=True,
    metrics=(
        MetricType.MSE,
        MetricType.ACCURACY,
        MetricType.PRECISION,
        MetricType.RECALL,
        MetricType.F1_SCORE
    ),
    update_on_error_only=False,
    mse_threshold=0.001,
    classification_threshold=0.5
)

plot_metric(history_lin, MetricType.MSE, "Table 2.9 - MSE")
plot_metric(history_lin, MetricType.ACCURACY, "Table 2.9 - Accuracy")

if len(x_lin[0]) == 2:
    plot_decision_boundary(lin_model, x_lin, y_lin, "Table 2.9 - Frontière de décision")

print_predictions(lin_model, x_lin, y_lin, classification=True, threshold=0.5)
```

# Part 4 : Régression linéaire (table 2.11)


```python
from neural_network import NeuralNetwork, Layer, TrainingType, ActivationFunction, MetricType
from neural_network.utils import (load_csv_dataset, plot_metric, plot_regression_result, print_predictions)

x_reg, y_reg = load_csv_dataset("data/table_2_11.csv", target_columns=1, has_header=False)

reg_model = NeuralNetwork()
reg_model.add_layer(Layer(n_inputs=len(x_reg[0]), n_neurons=1, activation=ActivationFunction.IDENTITY))

history_reg = reg_model.train(
    x_train=x_reg,
    y_train=y_reg,
    epochs=500,
    learning_rate=0.0001,
    strategy=TrainingType.STOCHASTIC,
    verbose=False,
    metrics=(MetricType.MSE,),
    update_on_error_only=False,
    mse_threshold=0.0001
)

plot_metric(history_reg, MetricType.MSE, "Table 2.11 - Régression - MSE")

y_reg_pred = [reg_model.predict(x) for x in x_reg]
plot_regression_result(x_reg, y_reg, y_reg_pred, "Table 2.11 - Régression linéaire")

print_predictions(reg_model, x_reg, y_reg, classification=False)
```
