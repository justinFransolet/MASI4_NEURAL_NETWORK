# Perceptron monocouche

## Classification à 3 classes (table 3.1)

```python
from neural_network import NeuralNetwork, Layer, TrainingType, ActivationFunction, MetricType
from neural_network.utils import (load_csv_dataset, plot_metric, print_multiclass_predictions, multiclass_accuracy, plot_multiclass_decision_boundaries_2d)

x_3c, y_3c = load_csv_dataset("data/table_3_1.csv", target_columns=3, has_header=False)

# conversion -1 / 1 vers 0 / 1
y_3c = [[0.0 if v == -1.0 else 1.0 for v in row] for row in y_3c]

model_3c = NeuralNetwork()
model_3c.add_layer(Layer(n_inputs=len(x_3c[0]), n_neurons=3, activation=ActivationFunction.IDENTITY))

history_3c = model_3c.train(
    x_train=x_3c,
    y_train=y_3c,
    epochs=500,
    learning_rate=0.001,
    strategy=TrainingType.STOCHASTIC,
    verbose=False,
    metrics=(MetricType.MSE,),
    update_on_error_only=False,
    mse_threshold=0.0001,
    classification_threshold=0.5
)

plot_metric(history_3c, MetricType.MSE, "Table 3.1 - MSE")
plot_multiclass_decision_boundaries_2d(model_3c, x_3c, y_3c, "Table 3.1 - Frontières de décision", threshold=0.5)

acc_3c = multiclass_accuracy(model_3c, x_3c, y_3c)
print(f"Accuracy multiclasses (table 3.1) = {acc_3c:.4f}")

print_multiclass_predictions(model_3c, x_3c, y_3c)
```

## Classification à 4 classes (table 3.5)

```python
from neural_network import NeuralNetwork, Layer, TrainingType, ActivationFunction, MetricType
from neural_network.utils import (load_csv_dataset, plot_metric, print_multiclass_predictions, multiclass_accuracy, plot_5x5_samples_with_predictions)

x_4c, y_4c = load_csv_dataset("data/table_3_5.csv", target_columns=4, has_header=False)

# conversion -1 / 1 vers 0 / 1
y_4c = [[0.0 if v == -1.0 else 1.0 for v in row] for row in y_4c]

model_4c = NeuralNetwork()
model_4c.add_layer(Layer(n_inputs=len(x_4c[0]), n_neurons=4, activation=ActivationFunction.IDENTITY))

history_4c = model_4c.train(
    x_train=x_4c,
    y_train=y_4c,
    epochs=500,
    learning_rate=0.01,
    strategy=TrainingType.STOCHASTIC,
    verbose=False,
    metrics=(MetricType.MSE,),
    update_on_error_only=False,
    mse_threshold=0.001,
    classification_threshold=0.5
)

plot_metric(history_4c, MetricType.MSE, "Table 3.5 - MSE")
plot_5x5_samples_with_predictions(model_4c, x_4c, y_4c, max_samples=12)

acc_4c = multiclass_accuracy(model_4c, x_4c, y_4c)
print(f"Accuracy multiclasses (table 3.5) = {acc_4c:.4f}")

print_multiclass_predictions(model_4c, x_4c, y_4c)
```
