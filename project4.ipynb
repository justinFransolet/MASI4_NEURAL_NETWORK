# Perceptron multicouche

## Opérateur logique XOR (table 4.3)

```python
from neural_network import NeuralNetwork, Layer, TrainingType, ActivationFunction, MetricType
from neural_network.utils import (
    plot_metric,
    plot_binary_decision_regions_2d,
    print_predictions
)

# XOR
x_xor = [[0, 0], [0, 1], [1, 0], [1, 1]]
y_xor = [[0], [1], [1], [0]]

xor_model = NeuralNetwork()
xor_model.add_layer(Layer(n_inputs=2, n_neurons=2, activation=ActivationFunction.SIGMOID))
xor_model.add_layer(Layer(n_inputs=2, n_neurons=1, activation=ActivationFunction.SIGMOID))

history_xor = xor_model.train(
    x_train=x_xor,
    y_train=y_xor,
    epochs=10000,
    learning_rate=0.5,
    strategy=TrainingType.STOCHASTIC,
    verbose=False,
    metrics=(
        MetricType.MSE,
        MetricType.ACCURACY,
        MetricType.PRECISION,
        MetricType.RECALL,
        MetricType.F1_SCORE
    ),
    update_on_error_only=False,
    mse_threshold=0.001,
    classification_threshold=0.5
)

plot_metric(history_xor, MetricType.MSE, "XOR - Évolution de la MSE")
plot_metric(history_xor, MetricType.ACCURACY, "XOR - Évolution de l'accuracy")

plot_binary_decision_regions_2d(
    xor_model,
    x_xor,
    y_xor,
    title="XOR - Régions de décision",
    threshold=0.5
)

print_predictions(xor_model, x_xor, y_xor, classification=True, threshold=0.5)
```

## Classification à 2 classes non linéairement séparable (table 4.12)

```python
from neural_network import NeuralNetwork, Layer, TrainingType, ActivationFunction, MetricType
from neural_network.utils import (
    load_csv_dataset,
    plot_metric,
    plot_binary_decision_regions_2d,
    print_predictions
)

x_412, y_412 = load_csv_dataset("data/table_4_12.csv", target_columns=1, has_header=False)

# Conversion éventuelle -1 / 1 vers 0 / 1

model_412 = NeuralNetwork()
model_412.add_layer(Layer(n_inputs=len(x_412[0]), n_neurons=20, activation=ActivationFunction.SIGMOID))
model_412.add_layer(Layer(n_inputs=20, n_neurons=1, activation=ActivationFunction.SIGMOID))

history_412 = model_412.train(
    x_train=x_412,
    y_train=y_412,
    epochs=10000,
    learning_rate=0.5,
    strategy=TrainingType.STOCHASTIC,
    verbose=False,
    metrics=(
        MetricType.MSE,
        MetricType.ACCURACY,
        MetricType.PRECISION,
        MetricType.RECALL,
        MetricType.F1_SCORE
    ),
    update_on_error_only=False,
    mse_threshold=0.001,
    classification_threshold=0.5
)

plot_metric(history_412, MetricType.MSE, "Table 4.12 - MSE")
plot_metric(history_412, MetricType.ACCURACY, "Table 4.12 - Accuracy")

if len(x_412[0]) == 2:
    plot_binary_decision_regions_2d(
        model_412,
        x_412,
        y_412,
        title="Table 4.12 - Régions de décision",
        threshold=0.5
    )

print_predictions(model_412, x_412, y_412, classification=True, threshold=0.5)
```

## Classification à 3 classes linéairement séparable (table 4.14)

```python
from neural_network import NeuralNetwork, Layer, TrainingType, ActivationFunction, MetricType
from neural_network.utils import (
    load_csv_dataset,
    plot_metric,
    print_multiclass_predictions,
    multiclass_accuracy,
    plot_multiclass_decision_regions_2d_with_uncertainty
)

x_414, y_414 = load_csv_dataset("data/table_4_14.csv", target_columns=3, has_header=False)

model_414 = NeuralNetwork()
model_414.add_layer(Layer(n_inputs=len(x_414[0]), n_neurons=20, activation=ActivationFunction.SIGMOID))
model_414.add_layer(Layer(n_inputs=20, n_neurons=3, activation=ActivationFunction.SIGMOID))

history_414 = model_414.train(
    x_train=x_414,
    y_train=y_414,
    epochs=20000,
    learning_rate=1.2,
    strategy=TrainingType.STOCHASTIC,
    verbose=False,
    metrics=(MetricType.MSE,),
    update_on_error_only=False,
    mse_threshold=0.001,
    classification_threshold=0.5
)

plot_metric(history_414, MetricType.MSE, "Table 4.14 - MSE")

plot_multiclass_decision_regions_2d_with_uncertainty(
    model_414,
    x_414,
    y_414,
    title="Table 4.14 - Régions de décision avec zone d'incertitude",
    step=0.025,
    margin=0.15
)

acc_414 = multiclass_accuracy(model_414, x_414, y_414)
print(f"Accuracy multiclasses (table 4.14) = {acc_414:.4f}")

print_multiclass_predictions(model_414, x_414, y_414)
```

## Régression non linéaire (table 4.17)

```python
from neural_network import NeuralNetwork, Layer, TrainingType, ActivationFunction, MetricType
from neural_network.utils import (
    load_csv_dataset,
    plot_metric,
    plot_regression_curve,
    print_predictions
)

x_417, y_417 = load_csv_dataset("data/table_4_17.csv", target_columns=1, has_header=False)

model_417 = NeuralNetwork()
model_417.add_layer(Layer(n_inputs=len(x_417[0]), n_neurons=8, activation=ActivationFunction.TANH))
model_417.add_layer(Layer(n_inputs=8, n_neurons=1, activation=ActivationFunction.IDENTITY))

history_417 = model_417.train(
    x_train=x_417,
    y_train=y_417,
    epochs=20000,
    learning_rate=0.01,
    strategy=TrainingType.STOCHASTIC,
    verbose=False,
    metrics=(MetricType.MSE,),
    update_on_error_only=False,
    mse_threshold=0.0005
)

plot_metric(history_417, MetricType.MSE, "Table 4.17 - Régression non linéaire - MSE")
plot_regression_curve(model_417, x_417, y_417, "Table 4.17 - Régression non linéaire")

print_predictions(model_417, x_417, y_417, classification=False)
```
